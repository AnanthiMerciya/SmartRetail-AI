```python
import pandas as pd

df = pd.read_csv("../dataset/processed/Superstore_Cleaned.csv")

# Annual Sales & Profit Matrix
yearly = df.groupby("Year").agg({"Sales": "sum", "Profit": "sum", "Quantity": "sum", "Order ID": "nunique"}).reset_index()
yearly["Margin"] = (yearly["Profit"] / yearly["Sales"]) * 100
print("Year-wise Breakdown:\n", yearly)
```

```python
# Category Breakdown
cat = df.groupby("Category").agg({"Sales": "sum", "Profit": "sum", "Quantity": "sum"}).reset_index()
cat["Margin"] = (cat["Profit"] / cat["Sales"]) * 100
print("Category Breakdown:\n", cat)
```

```python
# Discount Elasticity Analysis
disc = df.groupby("Discount").agg({"Sales": "sum", "Profit": "sum", "Quantity": "sum"}).reset_index()
disc["Margin"] = (disc["Profit"] / disc["Sales"]) * 100
print("Discount Sensitivity:\n", disc)
```
