```python
import pandas as pd

df = pd.read_csv("../dataset/processed/Superstore_Cleaned.csv")
df["Order Date"] = pd.to_datetime(df["Order Date"])

# Monthly Sales Resampling
ts = df.groupby(df["Order Date"].dt.to_period("M"))["Sales"].sum().reset_index()
ts.columns = ["Month", "Sales"]
ts["Month_Dt"] = pd.to_datetime(ts["Month"].astype(str))
ts["Year"] = ts["Month_Dt"].dt.year
ts["Month_Number"] = ts["Month_Dt"].dt.month

# Autoregressive Features: Lags & Rolling Window
ts["Lag_1"] = ts["Sales"].shift(1)
ts["Lag_2"] = ts["Sales"].shift(2)
ts["Lag_12"] = ts["Sales"].shift(12)
ts["Rolling_3"] = ts["Sales"].shift(1).rolling(3).mean()

print("Feature Matrix Sample:\n", ts.tail(12))
```
