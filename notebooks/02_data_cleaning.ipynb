```python
import os
import pandas as pd

df = pd.read_csv("../dataset/raw/Superstore.csv", encoding="latin1")
print("Initial records:", len(df))

# Deduplication
df = df.drop_duplicates()

# Date parsing
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Ship Date"] = pd.to_datetime(df["Ship Date"])

# Clean numeric attributes
df["Sales"] = pd.to_numeric(df["Sales"], errors="coerce").fillna(0.0)
df["Profit"] = pd.to_numeric(df["Profit"], errors="coerce").fillna(0.0)
df["Quantity"] = pd.to_numeric(df["Quantity"], errors="coerce").fillna(1).astype(int)
df["Discount"] = pd.to_numeric(df["Discount"], errors="coerce").fillna(0.0)

# Extract temporal attributes
df["Year"] = df["Order Date"].dt.year
df["Month"] = df["Order Date"].dt.month
df["Month Name"] = df["Order Date"].dt.month_name()
df["Quarter"] = df["Order Date"].dt.quarter

os.makedirs("../dataset/processed", exist_ok=True)
df.to_csv("../dataset/processed/Superstore_Cleaned.csv", index=False)
print("Processed dataset saved successfully. Final records:", len(df))
```
